```python
categories = [
    "Civil",
    "Structural",
    "Mechanical",
    "Electrical",
    "HVAC",
    "IT/Controls",
]


package_rows = []

for _, project in projects.iterrows():

    # Create 12 budget shares that together add up to 100%
    weights = np.random.dirichlet(
        np.ones(12)
    )

    for i in range(1, 13):

        category = categories[
            (i - 1) % len(categories)
        ]

        package_rows.append({
            "package_code":
                f"{project.project_code}-PKG-{i:02d}",

            "project_code":
                project.project_code,

            "package_name":
                f"{category} Package {i:02d}",

            "cost_category":
                category,

            "package_budget":
                round(
                    project.approved_budget
                    * weights[i - 1],
                    2,
                ),
        })


packages = pd.DataFrame(package_rows)

packages
```

```python
budget_check = (
    packages
    .groupby("project_code", as_index=False)["package_budget"]
    .sum()
    .rename(columns={"package_budget": "sum_of_package_budgets"})
)

budget_check = budget_check.merge(
    projects[
        ["project_code", "approved_budget"]
    ],
    on="project_code",
    how="left",
)

budget_check["difference"] = (
    budget_check["sum_of_package_budgets"]
    - budget_check["approved_budget"]
)

budget_check
```

```python
# Empty lists where we will collect our generated records
progress_rows = []
cost_rows = []


for _, project in projects.iterrows():

    # Convert the project's start date to the end of that month
    start = (
        pd.Timestamp(project.start_date)
        .to_period("M")
        .to_timestamp("M")
    )

    # We only generate data up to our reporting cut-off date
    end = min(
        REPORTING_END,
        pd.Timestamp(project.planned_end_date)
        .to_period("M")
        .to_timestamp("M"),
    )

    # All reporting months between project start and reporting cut-off
    months = pd.date_range(
        start,
        end,
        freq="ME",
    )

    # All months across the project's complete planned lifetime
    full_project_months = pd.date_range(
        start,
        pd.Timestamp(project.planned_end_date)
        .to_period("M")
        .to_timestamp("M"),
        freq="ME",
    )

    total_months = max(
        len(full_project_months),
        1,
    )

    # Find only the cost packages belonging to this project
    project_packages = packages[
        packages["project_code"] == project.project_code
    ].copy()

    print(
        project.project_code,
        "| reporting months:",
        len(months),
        "| total planned months:",
        total_months,
        "| packages:",
        len(project_packages),
    )
```

```python
progress_rows = []

for _, project in projects.iterrows():

    start = (
        pd.Timestamp(project.start_date)
        .to_period("M")
        .to_timestamp("M")
    )

    end = min(
        REPORTING_END,
        pd.Timestamp(project.planned_end_date)
        .to_period("M")
        .to_timestamp("M"),
    )

    months = pd.date_range(
        start,
        end,
        freq="ME",
    )

    full_project_months = pd.date_range(
        start,
        pd.Timestamp(project.planned_end_date)
        .to_period("M")
        .to_timestamp("M"),
        freq="ME",
    )

    total_months = max(
        len(full_project_months),
        1,
    )

    for month_number, month in enumerate(
        months,
        start=1,
    ):

        planned_progress = min(
            100,
            (
                month_number
                / total_months
            ) * 100,
        )

        actual_progress = min(
            100,
            max(
                0,
                planned_progress
                * project.schedule_health
                + np.random.normal(
                    0,
                    1.5,
                ),
            ),
        )

        progress_rows.append({
            "project_code":
                project.project_code,

            "report_month":
                month.strftime("%Y-%m-%d"),

            "planned_progress_pct":
                round(
                    planned_progress,
                    2,
                ),

            "actual_progress_pct":
                round(
                    actual_progress,
                    2,
                ),

            "reported_by":
                np.random.choice([
                    "PMO",
                    "Site Team",
                    "Project Controls",
                ]),

            "source_system":
                "P6_EXPORT",
        })


progress = pd.DataFrame(progress_rows)

progress
```

```python
cost_rows = []

for _, project in projects.iterrows():

    start = (
        pd.Timestamp(project.start_date)
        .to_period("M")
        .to_timestamp("M")
    )

    end = min(
        REPORTING_END,
        pd.Timestamp(project.planned_end_date)
        .to_period("M")
        .to_timestamp("M"),
    )

    months = pd.date_range(
        start,
        end,
        freq="ME",
    )

    full_project_months = pd.date_range(
        start,
        pd.Timestamp(project.planned_end_date)
        .to_period("M")
        .to_timestamp("M"),
        freq="ME",
    )

    total_months = max(
        len(full_project_months),
        1,
    )

    project_packages = packages[
        packages["project_code"] == project.project_code
    ].copy()

    for month_number, month in enumerate(
        months,
        start=1,
    ):

        monthly_weight = (
            (1 / total_months)
            * (
                0.65
                + 0.7
                * np.sin(
                    np.pi
                    * min(
                        month_number / total_months,
                        1,
                    )
                )
            )
        )

        for _, package in project_packages.iterrows():

            planned = (
                package.package_budget
                * monthly_weight
            )

            cost_noise = np.random.normal(
                1.0 / project.cost_health,
                0.08,
            )

            actual = max(
                0,
                planned * cost_noise,
            )

            committed = max(
                actual,
                planned
                * np.random.uniform(
                    1.00,
                    1.20,
                ),
            )

            forecast = (
                package.package_budget
                / project.cost_health
                * np.random.uniform(
                    0.98,
                    1.03,
                )
            )

            cost_rows.append({
                "cost_record_id":
                    f"C-{len(cost_rows)+1:06d}",

                "project_code":
                    project.project_code,

                "package_code":
                    package.package_code,

                "cost_category":
                    package.cost_category,

                "report_month":
                    month.strftime("%Y-%m-%d"),

                "planned_cost":
                    round(planned, 2),

                "actual_cost":
                    round(actual, 2),

                "committed_cost":
                    round(committed, 2),

                "forecast_at_completion":
                    round(forecast, 2),

                "currency":
                    "GBP",

                "source_system":
                    "ERP_COST",
            })


costs = pd.DataFrame(cost_rows)

costs
```

```python
activity_rows = []

activity_names = [
    "Design approval",
    "Site mobilisation",
    "Groundworks",
    "Structural frame",
    "MEP installation",
    "Electrical installation",
    "HVAC installation",
    "Testing & commissioning",
    "Quality inspection",
    "Client handover",
]

for _, project in projects.iterrows():

    project_start = pd.Timestamp(
        project.start_date
    )

    for i in range(1, 41):

        planned_start = (
            project_start
            + pd.Timedelta(
                days=(i - 1) * 12
                + np.random.randint(0, 8)
            )
        )

        duration = np.random.randint(
            10,
            46,
        )

        planned_finish = (
            planned_start
            + pd.Timedelta(
                days=duration
            )
        )

        delay_bias = max(
            0,
            (1 - project.schedule_health) * 90,
        )

        start_delay = max(
            0,
            int(
                np.random.normal(
                    delay_bias * 0.35,
                    4,
                )
            ),
        )

        finish_delay = max(
            0,
            int(
                np.random.normal(
                    delay_bias,
                    7,
                )
            ),
        )

        actual_start = (
            planned_start
            + pd.Timedelta(
                days=start_delay
            )
        )

        actual_finish = (
            planned_finish
            + pd.Timedelta(
                days=finish_delay
            )
        )

        if planned_start > REPORTING_END:

            status = "Not Started"
            percent_complete = 0
            actual_start_output = ""
            actual_finish_output = ""

        elif actual_finish <= REPORTING_END:

            status = "Complete"
            percent_complete = 100

            actual_start_output = (
                actual_start.strftime("%Y-%m-%d")
            )

            actual_finish_output = (
                actual_finish.strftime("%Y-%m-%d")
            )

        else:

            status = "In Progress"

            elapsed = (
                REPORTING_END
                - actual_start
            ).days

            percent_complete = round(
                max(
                    5,
                    min(
                        95,
                        elapsed
                        / max(
                            duration + finish_delay,
                            1,
                        )
                        * 100,
                    ),
                ),
                1,
            )

            actual_start_output = (
                actual_start.strftime("%Y-%m-%d")
            )

            actual_finish_output = ""

        activity_rows.append({
            "activity_id":
                f"{project.project_code}-ACT-{i:03d}",

            "project_code":
                project.project_code,

            "wbs_code":
                f"WBS-{((i - 1) // 5) + 1:02d}",

            "activity_name":
                activity_names[
                    (i - 1) % len(activity_names)
                ],

            "discipline":
                categories[
                    (i - 1) % len(categories)
                ],

            "planned_start":
                planned_start.strftime("%Y-%m-%d"),

            "planned_finish":
                planned_finish.strftime("%Y-%m-%d"),

            "actual_start":
                actual_start_output,

            "actual_finish":
                actual_finish_output,

            "percent_complete":
                percent_complete,

            "status":
                status,

            "is_critical":
                bool(
                    i % 7 == 0
                    or i % 11 == 0
                ),

            "source_system":
                "P6_EXPORT",
        })


schedule = pd.DataFrame(activity_rows)

schedule
```

```python
procurement_rows = []


for _, package in packages.iterrows():

    project = projects[
        projects["project_code"]
        == package.project_code
    ].iloc[0]

    # Most cost packages will have 1 procurement order.
    # Some will have 2.
    number_of_orders = np.random.choice([
        1,
        1,
        1,
        2,
    ])


    for _ in range(number_of_orders):

        # Request for quotation date
        rfq_date = (
            pd.Timestamp(
                project.start_date
            )
            + pd.Timedelta(
                days=np.random.randint(
                    20,
                    240,
                )
            )
        )

        # Purchase order date
        po_date = (
            rfq_date
            + pd.Timedelta(
                days=np.random.randint(
                    10,
                    50,
                )
            )
        )

        # Planned delivery date
        planned_delivery = (
            po_date
            + pd.Timedelta(
                days=np.random.randint(
                    30,
                    120,
                )
            )
        )

        # Projects with worse schedule health
        # are more likely to experience procurement delays
        delay = max(
            0,
            int(
                np.random.normal(
                    max(
                        0,
                        (
                            1
                            - project.schedule_health
                        ) * 80,
                    ),
                    9,
                )
            ),
        )

        actual_delivery = (
            planned_delivery
            + pd.Timedelta(
                days=delay
            )
        )


        # Decide current procurement status
        if planned_delivery > REPORTING_END:

            status = "Ordered"
            actual_delivery_output = ""

        elif actual_delivery <= REPORTING_END:

            status = "Delivered"

            actual_delivery_output = (
                actual_delivery.strftime(
                    "%Y-%m-%d"
                )
            )

        else:

            status = "Delayed"
            actual_delivery_output = ""


        procurement_rows.append({

            "procurement_id":
                f"PO-{len(procurement_rows)+1:05d}",

            "project_code":
                package.project_code,

            "package_code":
                package.package_code,

            "vendor_id":
                np.random.choice(
                    vendors.vendor_id
                ),

            "category":
                package.cost_category,

            "rfq_date":
                rfq_date.strftime(
                    "%Y-%m-%d"
                ),

            "po_date":
                po_date.strftime(
                    "%Y-%m-%d"
                ),

            "planned_delivery_date":
                planned_delivery.strftime(
                    "%Y-%m-%d"
                ),

            "actual_delivery_date":
                actual_delivery_output,

            "package_value":
                round(
                    package.package_budget
                    * np.random.uniform(
                        0.35,
                        0.85,
                    ),
                    2,
                ),

            "status":
                status,

            "criticality":
                np.random.choice(
                    [
                        "Low",
                        "Medium",
                        "High",
                    ],
                    p=[
                        0.25,
                        0.50,
                        0.25,
                    ],
                ),

            "source_system":
                "PROCURE_TRACKER",
        })


procurement = pd.DataFrame(
    procurement_rows
)

procurement
```

```python
# ============================================================
# INTENTIONAL RAW DATA QUALITY PROBLEMS
# ============================================================

# 1. Missing vendor IDs
bad_vendor_rows = procurement.sample(
    frac=0.025,
    random_state=SEED,
).index

procurement.loc[
    bad_vendor_rows,
    "vendor_id",
] = np.nan


# 2. Duplicate procurement record
procurement = pd.concat(
    [
        procurement,
        procurement.iloc[[5]],
    ],
    ignore_index=True,
)


# 3. Invalid / non-existent project reference
costs.loc[
    costs.index[17],
    "project_code",
] = "PRJ-999"


# 4. Negative actual cost
costs.loc[
    costs.index[80],
    "actual_cost",
] = -25_000


# 5. Extreme cost anomaly
costs.loc[
    costs.index[120],
    "actual_cost",
] *= 8


# 6. Missing package code
costs.loc[
    costs.index[180],
    "package_code",
] = np.nan


# 7. Inconsistent date format
progress.loc[
    progress.index[9],
    "report_month",
] = "31/12/2025"


# 8. Impossible progress percentage
progress.loc[
    progress.index[25],
    "actual_progress_pct",
] = 126.0


# 9. Actual finish before actual start
schedule_row = schedule.index[30]

schedule.loc[
    schedule_row,
    "actual_start",
] = "2026-06-20"

schedule.loc[
    schedule_row,
    "actual_finish",
] = "2026-06-10"


# 10. Missing completion percentage
schedule.loc[
    schedule.index[65],
    "percent_complete",
] = np.nan


print("Intentional data-quality problems added successfully.")
```

```python
# ============================================================
# EXPORT RAW DATASETS
# ============================================================

# Save projects WITHOUT the simulation helper columns
projects.drop(
    columns=[
        "cost_health",
        "schedule_health",
    ]
).to_csv(
    RAW / "projects.csv",
    index=False,
)

# Save vendors
vendors.to_csv(
    RAW / "vendors.csv",
    index=False,
)

# Save cost packages
packages.to_csv(
    RAW / "cost_packages.csv",
    index=False,
)

# Save monthly costs
costs.to_csv(
    RAW / "cost_monthly.csv",
    index=False,
)

# Save schedule activities
schedule.to_csv(
    RAW / "schedule_activities.csv",
    index=False,
)

# Save procurement data
procurement.to_csv(
    RAW / "procurement_packages.csv",
    index=False,
)

# Save progress data
progress.to_csv(
    RAW / "progress_updates.csv",
    index=False,
)

print("All raw datasets exported successfully.")
```
